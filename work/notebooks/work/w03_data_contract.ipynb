```python
from datasets import load_dataset

daily = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train"
)

print(daily)
print(daily.column_names)
```

```python
daily_df = daily.to_pandas()

daily_df["report_date"] = pd.to_datetime(daily_df["report_date"])

march = daily_df[
    (daily_df["report_date"] >= "2026-03-01") &
    (daily_df["report_date"] <= "2026-03-31")
]

print(march.shape)
march.head()
```

# Search Intelligence Data Contract

### 1. What does one row represent?
One row represents the daily performance of one content page for one client on one reporting date.

### 2. Which table(s) will you use?
I use:
- dim_content
- fact_content_daily_performance

### 3. Which time window?
March 2026 (2026-03-01 to 2026-03-31).

### 4. What am I predicting?
I aim to predict or rank content performance using historical Search Console and Analytics metrics.

### 5. What do I deliberately exclude?
Future performance metrics and any label-derived information are excluded to prevent data leakage.

```python
grain = march.groupby(
    ["report_date", "client_hash_id", "content_hash_id"]
).size()

print("Maximum duplicate rows:", grain.max())
```

Result:
Each row uniquely represents one client, one content page, and one reporting date.

```python
print("Rows:", len(march))
print("First Date:", march["report_date"].min())
print("Last Date:", march["report_date"].max())
```

```python
available = march[
    (march["gsc_data_available"] == True) &
    (march["ga4_data_available"] == True)
]

print("Available rows:", len(available))
```

Only rows where both Search Console and GA4 data are available are used.

```python
features = available[
    [
        "gsc_impressions",
        "gsc_clicks",
        "gsc_avg_position",
        "ga4_sessions",
        "scroll_events"
    ]
]

features.head()
```

## Feature Availability

1. gsc_impressions
Available because impressions are already recorded before prediction.

2. gsc_clicks
Available because clicks are historical observations.

3. gsc_avg_position
Available because search ranking history already exists.

4. ga4_sessions
Available because website sessions are historical analytics data.

5. scroll_events
Available because user engagement has already occurred.

```python
leak = available[
    [
        "gsc_impressions",
        "gsc_clicks",
        "ga4_sessions"
    ]
].copy()

# Deliberately leaking the label
leak["label"] = available["gsc_clicks"]

print(leak.head())
```

The column 'label' is directly derived from the prediction target. Including it would produce unrealistically high model performance because the answer is already present in the features. This demonstrates target leakage. The leaked feature should be removed before training.

## Limitation

This analysis uses only March 2026 and therefore does not capture seasonal variations or long-term trends. Some analytics fields also contain missing values, limiting the available training data.
